```python
movie_data = df[df["type"] == "Movie"].copy()

movie_data["duration_minutes"] = (
    movie_data["duration"].str.replace(" min", "", regex=False)
)

movie_data["duration_minutes"] = pd.to_numeric(
    movie_data["duration_minutes"], errors="coerce"
)

movie_data[["release_year", "duration_minutes"]].head()
```

```python
plt.scatter(movie_data["release_year"], movie_data["duration_minutes"])

plt.title("Release Year vs Movie Duration")
plt.xlabel("Release Year")
plt.ylabel("Duration (Minutes)")
plt.show()
```

```python
content_counts = df["type"].value_counts()

plt.pie(
    content_counts,
    labels=content_counts.index,
    autopct="%1.1f%%"
)

plt.title("Movies vs TV Shows on Netflix")
plt.show()
```

```python
movie_data[["release_year", "duration_minutes"]].corr()
```

```python
correlation = movie_data[["release_year", "duration_minutes"]].corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```
